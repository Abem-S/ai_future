```python
from qiskit import QuantumCircuit, transpile
from qiskit_aer import Aer
from qiskit.visualization import plot_histogram
from collections import Counter
```

```python
# --- 1. Define the Quantum Circuit ---
# Create a quantum circuit with 1 qubit (q) and 1 classical bit (c)
qc = QuantumCircuit(1, 1)

print("Quantum Circuit created with 1 qubit and 1 classical bit.")
```

```python
# --- 2. Initialize Superposition ---
# Apply the Hadamard gate (H) to the qubit (index 0).
# The H-gate puts the qubit into a state of superposition (simultaneously 0 and 1).
qc.h(0)
print("\nHadamard gate applied: Qubit is now in superposition.")
# Draw the circuit to see the gates applied.
print(qc.draw(output='text'))
```

```python
# --- 3. Measurement ---
# Measure the qubit (index 0) and store the result in the classical bit (index 0).
qc.measure(0, 0)
```

```python
# --- 4. Simulation and Execution (using Aer simulator) ---
# Use Qiskit's built-in simulator for testing the circuit locally.
# We ensure the backend is retrieved safely.
try:
    simulator = Aer.get_backend('qasm_simulator')
except Exception as e:
    print(f"\nERROR: Could not find Aer simulator. Ensure 'qiskit-aer' is installed and runtime is restarted. Details: {e}")
    # Exit gracefully if the simulator isn't found
    exit()

# Transpile (optimize) the circuit for the target backend
compiled_circuit = transpile(qc, simulator)

# Run the simulation 1024 times (shots) to see the probability distribution
shots = 1024
job = simulator.run(compiled_circuit, shots=shots)
result = job.result()

# Get the probability counts (how many times '0' and '1' were measured)
counts = result.get_counts(qc)
```

```python
# --- 5. Output Results ---
print(f"\nSimulation Results (Total Shots: {shots}):")
print(counts)

# Calculate probabilities (should be close to 50% for 0 and 50% for 1)
prob_0 = counts.get('0', 0) / shots * 100
prob_1 = counts.get('1', 0) / shots * 100

print(f"\nProbability of measuring 0: {prob_0:.2f}%")
print(f"Probability of measuring 1: {prob_1:.2f}%")

print("\n--- Execution Complete ---")
```
